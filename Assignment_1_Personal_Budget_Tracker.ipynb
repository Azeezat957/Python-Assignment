```python
#Building a script that tracks one month of personal spending against a budget for each category
```

```python
'''Listing Expenses '''

expenses = ( {'category': 'Food', 'item' : 'Spagetti', 'amount': 20000},
        {'category': 'Transport Fare', 'item' : 'Uber', 'amount': 30000},
         {'category': 'Expenses', 'item' : 'Utility Fee', 'amount': 50000},
        {'category': 'Expenses', 'item' : 'Friendship date', 'amount': 80000},
        {'category': 'Airtime', 'item' : 'Internet', 'amount': 19000},
        {'category': 'Expenses', 'item' : 'Music Subcription', 'amount': 10000},
        {'category': 'Expenses', 'item' : 'Canva pro payment', 'amount': 5000})


             
```

```python
'''Set up Budgets'''

budgets = { 'Food' : 30000,
           'Transport Fare': 30000,
            'Expenses' : 60000,
            'Airtime' : 19000 }
            
            
```

```python
''' Calculating spending per category '''

category_spending = {}

for expense in expenses:
    category = expense['category']
    amount = expense['amount']

    if category in category_spending:
        category_spending[category] += amount
    else:
        category_spending[category] = amount

print(category_spending)
```

```python
'''Comparing Spending With the Budget'''

for category, spent in category_spending.items ():
     
     budget = budgets[category]

     if spent > budget :
         status = 'OVER BUDGET'
     elif spent == budget:
         status = 'ON BUDGET' 
     else :
           status = 'WITHIN BUDGET'

     print( category, status)       
             
```

```python
'''A clear Summary'''


print("PERSONAL BUDGET TRACKER")
print("------------------------")

for category, spent in category_spending.items():

    budget = budgets[category]

    if spent > budget:
        status = "OVER BUDGET"
    elif spent == budget:
        status = "ON BUDGET"
    else:
        status = "WITHIN BUDGET"

    print(f"{category}: Spent ₦{spent:,} of ₦{budget:,} - {status}")
```
